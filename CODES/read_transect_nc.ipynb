```python
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from sklearn.metrics import r2_score

# Folder containing runup NetCDFs
nc_dir = Path("/path/to/netcdfs")

calc_2perc = []
cirn_2perc = []
filenames = []

for ncfile in sorted(nc_dir.glob("*.nc")):

    try:
        ds = xr.open_dataset(ncfile)

        # TWL timeseries for this burst
        twl = ds["TWL"].values

        # Remove NaNs
        twl = twl[np.isfinite(twl)]

        if len(twl) == 0:
            continue

        # Empirical 2% exceedance
        twl_2_calc = np.percentile(twl, 98)

        # CIRN value
        twl_2_cirn = float(ds["TWLstats_2exceedence_peaks"].values)

        if np.isfinite(twl_2_calc) and np.isfinite(twl_2_cirn):
            calc_2perc.append(twl_2_calc)
            cirn_2perc.append(twl_2_cirn)
            filenames.append(ncfile.name)

        ds.close()

    except Exception as e:
        print(f"Skipping {ncfile.name}: {e}")

calc_2perc = np.array(calc_2perc)
cirn_2perc = np.array(cirn_2perc)

# Statistics
r2 = r2_score(cirn_2perc, calc_2perc)
bias = np.mean(calc_2perc - cirn_2perc)
rmse = np.sqrt(np.mean((calc_2perc - cirn_2perc)**2))

# Plot
plt.figure(figsize=(8,8))

plt.scatter(
    cirn_2perc,
    calc_2perc,
    alpha=0.7
)

mn = min(cirn_2perc.min(), calc_2perc.min())
mx = max(cirn_2perc.max(), calc_2perc.max())

plt.plot(
    [mn, mx],
    [mn, mx],
    'k--',
    lw=2,
    label='1:1'
)

plt.xlabel('TWLstats_2exceedence_peaks')
plt.ylabel('98th Percentile of TWL')
plt.title(f'N = {len(calc_2perc)} bursts')

plt.text(
    0.05,
    0.95,
    f'R² = {r2:.3f}\nBias = {bias:.3f} m\nRMSE = {rmse:.3f} m',
    transform=plt.gca().transAxes,
    va='top'
)

plt.legend()
plt.tight_layout()
plt.show()
```

```python
import glob
import os
import re

import numpy as np
import pandas as pd
import xarray as xr
import netcdf4 as nc

dirpath = '/home/cvolpano/runup-Chelsea/DATA/DATA/results/netcdf/'
records = []
os.chdir(dirpath)

for file in glob.glob("*.nc"):
    fname = os.path.basename(file)
    transect = int(re.search(r'transect(\d+)', fname).group(1))
    
    camera = re.search(r'\.(c\d)\.', fname).group(1)
    ds = xr.open_dataset(file)
    ds = ds.sel(T_dim = slice("2025-01-23", "2025-02-19"))
  
    t = (ds["T_dim"].values)
    
    zrunup = (ds["Zrunup"].values)
   
    beta06 = ds['TWLstats_betaS2006'].values
    twl_runup = ds['twl_runup'].values
    TWL = ds["TWL"].values
    twl = ds["twl"].values
    twl_hs = ds["twl_hs"].values
    for ti, ri in zip(t, zrunup):
        records.append({
            "time": ti,
            "transect": transect,
            "camera": camera,
            "Zrunup": zrunup,
            "twl_runup": twl_runup,
            "TWL": TWL,
            "twl": twl,
            "Hs": twl_hs
        })

keys = list(ds.data_vars.keys())
df = pd.DataFrame(records)

df = df[(df["transect"]==20)]


plt.figure(figsize=(12,6))
ds["Zrunup"].plot(
    x="T_dim",
)
plt.show()
```
